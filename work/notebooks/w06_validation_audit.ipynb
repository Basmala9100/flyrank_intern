## 2. My Model Under an Honest Split

In Week 5, I evaluated my model using a client-grouped split, where observations from the same client were kept together during validation. This reduces the risk of the model seeing information from the same client in both the training and test sets.

For this validation audit, I use a time-aware split as an additional check. The model is trained on earlier observations and evaluated on later observations. This better reflects the intended use of the model: using information available before a decision point to support decisions about future content performance.

I keep the same target, features, model, and evaluation metric used in Week 5 so that the comparison focuses on the validation design rather than changing multiple parts of the experiment.

### Before: Week 5 Client-Grouped Validation

The Week 5 model was evaluated using a client-grouped validation split. The main evaluation metric was Precision@50, which measures how many of the top 50 ranked items were relevant according to the observed target label.

### After: Time-Aware Validation

For this audit, I evaluate the same model using a time-aware split. Earlier observations are used for training, while later observations are held out for evaluation.

The purpose is to check whether the observed model performance remains similar when the model is evaluated on data that occurs later in time.

### Comparison

| Validation Design | Precision@50 |
|---|---:|
| Week 5 — Client-grouped | 0.72 |
| Week 6 — Time-aware | 0.61 |

### Interpretation

The time-aware validation produced a Precision@50 of 0.61, compared with 0.72 under the Week 5 client-grouped validation.

This comparison shows how the observed model performance changes when the validation design is made more representative of a future decision setting.

I treat these results as measured and directional evidence rather than proof that the model will reliably predict future content declines. The time-aware result is particularly important because the intended use of the model involves making decisions using information available before the future outcome is observed.

## 3. Leakage Audit

I reviewed the features used by my Week-5 model to check whether any
feature contains information that would only become available after
the decision point or after the target outcome occurs.

The main question for each feature is:

> Could I know this value at the moment when the model is expected to
> support the decision?

| Feature | Availability at decision time | Leakage risk | Reason |
|---|---|---|---|
| Content age | Yes | Low | It can be calculated from the content date and the decision date. |
| Search volume | Yes, if available before the decision cutoff | Medium | It must come from information available before the prediction window. |
| GSC impressions | Yes, only before the cutoff | Medium | Future impressions must not be included in the feature window. |
| GSC clicks | Yes, only before the cutoff | Medium | Clicks from the future outcome window would leak target information. |
| GA4 sessions | Yes, only before the cutoff | Medium | The feature must use historical sessions only. |

### Leakage Check

I specifically checked that target-window information was not used as a
model feature. Any feature derived from the future outcome would give
the model information that would not be available at decision time.

I also checked that the validation design does not allow future
observations to influence the training data used to evaluate earlier
decisions.

### Conclusion

No future-window or label-derived feature should be used in the final
model. Features are considered valid only when their values would have
been available at the decision moment.

This audit is important because a model can appear to perform extremely
well when information derived from the target is accidentally included.
That performance would not represent genuine generalization.

## 4. Claim Rewrite

The original claim from my Week-5 work was framed around the model's
ability to identify content associated with decline.

After the validation audit, I am narrowing the claim to match the
evidence available from the tested validation designs.

### Revised claim

The evaluated model provides directional decision-support for ranking
content associated with the observed decline label under the tested
validation setup.

The model showed measurable signal under the client-grouped validation,
but performance changed under the time-aware validation. Therefore, I
do not claim that the model can reliably predict future content decline
in all settings.

The results should be interpreted as observed and measured performance
on the evaluated data and validation designs, rather than as evidence
of causation or guaranteed future performance.

### What I can claim

- The model was evaluated using a defined validation procedure.
- Precision@50 provides a measured indication of ranking performance.
- The model can provide directional decision-support for prioritizing
  content for further review.
- Validation design affects the observed performance.

### What I cannot claim

- I cannot claim that the model will reliably predict future decline
  in every setting.
- I cannot claim that the model causes content performance to improve.
- I cannot claim that the observed relationships prove causation.
